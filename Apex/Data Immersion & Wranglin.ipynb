```python
# Convert Order_Date into datetime format

ex["Order_Date"] = pd.to_datetime(
    ex["Order_Date"],
    errors="coerce"
)

# Check the data type
print(ex["Order_Date"].dtype)
```

```python
# Check invalid dates
print("Invalid Dates:",
      ex["Order_Date"].isnull().sum())

# Check date range
print("Minimum Date:",
      ex["Order_Date"].min())

print("Maximum Date:",
      ex["Order_Date"].max())
```

```python
ex["Age"] = ex["Age"].astype("Int64")

print(ex["Age"].dtype)
```

```python
# Calculate expected sales
ex["Calculated_Sales"] = (
    ex["Quantity"] * ex["Unit_Price"]
).round(2)

# Compare with actual sales
ex["Sales_Mismatch"] = (
    (ex["Total_Sales"] - ex["Calculated_Sales"]).abs() > 0.01
)

print("Sales Mismatch:")
print(ex["Sales_Mismatch"].sum())
```

```python
# Extract year, month and day
ex["Order_Year"] = ex["Order_Date"].dt.year
ex["Order_Month"] = ex["Order_Date"].dt.month
ex["Order_Day"] = ex["Order_Date"].dt.day

# Categorize customers by age
ex["Age_Group"] = pd.cut(
    ex["Age"],
    bins=[0, 25, 40, 60, 100],
    labels=["Young", "Adult", "Middle_Aged", "Senior"]
)

# Calculate total sales independently
ex["Calculated_Sales"] = (
    ex["Quantity"] * ex["Unit_Price"]
).round(2)


# Display newly created columns
ex[[
    "Order_Date",
    "Order_Year",
    "Order_Month",
    "Order_Day",
    "Age",
    "Age_Group",
    "Quantity",
    "Unit_Price",
    "Calculated_Sales"
]].head(10)
```

```python
print("New Columns:")
print(ex.columns.tolist())

print("\nDataset Shape:")
print(ex.shape)

print("\nAge Group Counts:")
print(ex["Age_Group"].value_counts())
```

```python
# Outlier detection using IQR
columns = ["Age", "Quantity", "Unit_Price", "Total_Sales"]

for col in columns:
    Q1 = ex[col].quantile(0.25)
    Q3 = ex[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = ex[
        (ex[col] < lower) | (ex[col] > upper)
    ]

    print(f"\n{col}")
    print("Lower Limit:", round(lower, 2))
    print("Upper Limit:", round(upper, 2))
    print("Outliers:", len(outliers))
```

```python
# Find Total_Sales outliers
Q1 = ex["Total_Sales"].quantile(0.25)
Q3 = ex["Total_Sales"].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

# Filter outliers
sales_outliers = ex[
    (ex["Total_Sales"] < lower) |
    (ex["Total_Sales"] > upper)
]

# Display outlier records
print("Total Outliers:", len(sales_outliers))

sales_outliers[[
    "Order_ID",
    "Customer_ID",
    "Product",
    "Category",
    "Quantity",
    "Unit_Price",
    "Total_Sales"
]].sort_values("Total_Sales", ascending=False)
```

```python
# Flag potential sales outliers
Q1 = ex["Total_Sales"].quantile(0.25)
Q3 = ex["Total_Sales"].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

ex["Sales_Outlier"] = (
    (ex["Total_Sales"] < lower) |
    (ex["Total_Sales"] > upper)
)

# Check the result
print(ex["Sales_Outlier"].value_counts())
```

```python
# Final validation
print("Dataset Shape:", ex.shape)

print("\nMissing Values:")
print(ex.isnull().sum().sum())

print("\nExact Duplicate Rows:")
print(ex.duplicated().sum())

print("\nSales Mismatch:")
print(ex["Sales_Mismatch"].sum())

print("\nSales Outliers:")
print(ex["Sales_Outlier"].sum())

print("\nDuplicate Order ID Rows:")
print(ex["Duplicate_Order_ID"].sum())
```

```python
# Remove temporary validation columns
df_final = ex.drop(
    columns=[
        "Calculated_Sales",
        "Sales_Mismatch"
    ],
    errors="ignore"
)

# Save final cleaned dataset
df_final.to_csv(
    "cleaned_sales.csv",
    index=False
)

print("Cleaned dataset saved successfully!")
print("Final Shape:", df_final.shape)
```
